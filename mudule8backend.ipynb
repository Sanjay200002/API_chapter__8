Question:-  Explain the core principal of restful APIs

answer :-

REST is stand for (Representional state transfer) is an architectural style. It is used for designing and networked applications.

in Core principal of restful APIs including client-server architecture, stateless communication, resource-based interaction, uniform interface, and cacheability.

1. :- The client-server architecture seprates the client(requesting resource) from the server(providing resource).
2. :- Stateless communications means that RESTful system do not store any client context on the server between requset .
3. :- Resource-based inttraction uses resource that can be access and manipulate through the url, with clients intracting using standard http method like GET, POST, PUT, AND DELETE.
4. :- Uniform interface means standard http methods , in uniform interface the url and media types are used to exchange data.
5. :- cacheability allow response to be cached to improve performance.


Question :- What are the common HTTP methods used in REST APIs and what are their typical uses?

Answer :-

Common HTTP methods are here:-

. GET
. POST
. PATCH
. PUT
. DELETE

1. :- GET :-
purpose:- Retrive data from the server.
behavior:- calling it multiple times and gives the same result.

Example:-
GET / users -> returns a list of users
GET / users/123 -> return details of user with id 123

2. :- POST :-
Purpose:- Create a new resource on the server.
behavior:-Repeated request may create duplicates.

example:-
POST / users with JSON body {"name" : "jeebachh"} ->
creaate a new user like {"id": 101, "name" : "jeebachh"}
with new id.



3. PATCH :-
Purpose :- partially update an existing resource.
behavior:- depend on implementation.

example:- PATCH / users/ 123 with {"age" : 24} -> update the user's age.
like {"id" : 123, "name" : "jeebachh", "age" : 24}   here age is updated


4. PUT :-
Purpose:- Update or replace an exixting resource completely.
behavior :- Repeated requests with the same data results in the same state.

example:- PUT /users/123 with body {"name" : "jeebachh", "age" : 24} -> replace user 123 data entirly.
like {"id" : 123, "name" : "jeebachh", "age" :24}     here replaced in the same state.


5. DELETE:-
Purpose:- Remove a resource from the server.
behavior:- Delete multiple times on the same resource has the same effect.

example:- DELETE /users/123 -> removes user 123  
here answer is no content because id 123 is deleted.

Question:- Explain the different between stateless stateful REST APIs.


Stateless Rest APIs :-

The server does not store any client session information between request.
Each request contains all the necessary information for the server to process it.

for example :- a weather application is a stateless API.


Stateful Rest APIs :-

The server stores client session data between requests.
The client and server maintain a continuous conversation where the server remembers the clients state (like shoping cart, login session etc).
